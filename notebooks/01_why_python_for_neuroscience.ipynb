## Why this matters

Almost every question you can ask about the brain ends up as a number in a file. A spike train is a list of times. An EEG recording is a long row of voltages sampled a few hundred times a second. A connectivity matrix is a grid of correlations between brain regions.

None of that fits in a spreadsheet, and none of it is one number you can eyeball. You need something that can read thousands of values, do the same operation to all of them, and draw the result. That is what we are here for.

## Three lines, one real brain signal

Before any explanation, here is the kind of thing you will be doing on your own by session 12. This is a real fMRI signal from one brain region of one person watching a movie inside a scanner.

Do not worry about the syntax yet. Just run it.

```python
# Where the course data lives.
# Locally the notebook finds it in ../data/. On Colab it pulls the same file from GitHub.
import os

REPO_RAW = "https://raw.githubusercontent.com/saeedrafsharx/python-for-neuroscience/main/data/"
DATA = "../data/" if os.path.exists("../data/one_subject_timeseries.csv") else REPO_RAW
print("reading data from:", DATA)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

signal = pd.read_csv(DATA + "one_subject_timeseries.csv")
signal.plot(x="time_s", y="Med DMN", figsize=(11, 3), legend=False)
plt.xlabel("time (seconds)")
plt.ylabel("signal (z-scored)")
plt.title("Medial default mode network, one participant")
plt.show()
```

That wiggly line is activity in the medial part of the default mode network, one of the best studied systems in the human brain. Every bump is roughly 2 seconds of scanner time.

Five lines of code, real data, a publishable-looking figure. That is the whole pitch.

## Why Python specifically

You could do a lot of this in MATLAB, and plenty of good labs still do. Python won on a different axis: it is free, it runs anywhere, and the neuroscience community has built an enormous amount of open tooling on top of it.

Here is a rough count of widely used open neuroscience toolboxes, grouped by the language you write them in. This is illustrative, not a survey.

```python
import matplotlib.pyplot as plt

tools = {
    "Python": ["MNE", "Nilearn", "NiBabel", "Brian2", "NEST (Python API)",
               "DIPY", "SpikeInterface", "Elephant", "Neo", "PyNN"],
    "MATLAB": ["SPM", "EEGLAB", "FieldTrip", "Brainstorm", "CONN"],
    "R": ["brainGraph", "eegkit"],
    "C / C++ core": ["FSL", "AFNI", "FreeSurfer"],
}

names = list(tools)
counts = [len(tools[n]) for n in names]

fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(names, counts, color=["#2b6cb0", "#a0aec0", "#a0aec0", "#a0aec0"])
ax.set_ylabel("number of well-known toolboxes", fontsize=13)
ax.set_title("Open neuroscience software, by main language", fontsize=15)
ax.tick_params(labelsize=12)
for i, c in enumerate(counts):
    ax.text(i, c + 0.15, str(c), ha="center", fontsize=13)
plt.tight_layout()
plt.show()

for lang, pkgs in tools.items():
    print(f"{lang:14s} {', '.join(pkgs)}")
```

The point is not that the bar is taller. The point is that whatever your data is, spikes or EEG or MRI or a modelled network, someone has already written the Python package for it, and you can read its source.
