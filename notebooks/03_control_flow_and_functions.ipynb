## Why this matters

Last session we wrote out four neurons by hand. Real recordings have hundreds of neurons and hundreds of trials, so hand-writing stops working immediately.

Three tools fix that. `if` decides, `for` repeats, and `def` packages something up so you can use it again without copying it.

## Making a decision with `if`

Suppose you want to label a neuron as quiet, active or bursting depending on its firing rate. In words that is a chain of conditions, and Python writes it almost the same way.

Notice the colon at the end of each line, and notice that the indented block belongs to the condition above it. Indentation is not decoration in Python. It is the syntax.

```python
rate = 12.5

if rate < 1:
    label = "quiet"
elif rate < 20:
    label = "active"
else:
    label = "bursting"

print(f"A neuron at {rate} Hz is {label}")
```

The comparisons available to you are `<`, `>`, `<=`, `>=`, `==` for "is equal to", and `!=` for "is not equal to". Two equals signs for comparison, one for assignment. That trips up everybody once.

You can also combine conditions with `and`, `or` and `not`.

```python
rate = 12.5
is_visual = True

if rate > 10 and is_visual:
    print("Fast-firing visual neuron")
```

## Repeating with `for`

A `for` loop takes a collection and hands you one item at a time. Here we have spike counts from eight trials of the same experiment.

```python
spike_counts = [12, 15, 9, 22, 18, 7, 25, 14]
duration_s = 2.0

for count in spike_counts:
    print(count / duration_s)
```

That printed the rates but threw them away. Usually you want to keep them, so you make an empty list and append to it as you go. This pattern shows up in every analysis script you will ever write.

```python
rates = []

for count in spike_counts:
    rates.append(count / duration_s)

print(rates)
```

If you also want the trial number, wrap the list in `enumerate()`. It hands you the position and the value together.

```python
for trial, count in enumerate(spike_counts):
    print(f"Trial {trial}: {count} spikes")
```

## Packaging it with `def`

Right now the formula `count / duration` is scattered through your code. Put it in a function once, and you can trust it everywhere.

A function takes inputs, does something, and hands a value back with `return`.

```python
def firing_rate(spike_count, duration):
    """Return spikes per second."""
    return spike_count / duration


print(firing_rate(250, 20.0))
print(firing_rate(9, 2.0))
```

Functions can hold logic too. Here is the classification from earlier, written once and reusable forever.

```python
def classify(rate):
    """Label a firing rate as quiet, active or bursting."""
    if rate < 1:
        return "quiet"
    elif rate < 20:
        return "active"
    else:
        return "bursting"


for count in spike_counts:
    r = firing_rate(count, duration_s)
    print(f"{r:5.1f} Hz  ->  {classify(r)}")
```

## The visual

Everything above, in one figure: rate per trial, with the classification thresholds drawn on top.

```python
import matplotlib.pyplot as plt

rates = [firing_rate(c, duration_s) for c in spike_counts]
trials = list(range(len(rates)))

fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(trials, rates, marker="o", linewidth=2, color="#2b6cb0", label="firing rate")
ax.axhline(1, color="grey", linestyle="--", label="quiet threshold")
ax.axhline(10, color="crimson", linestyle="--", label="bursting threshold")
ax.set_xlabel("trial", fontsize=13)
ax.set_ylabel("firing rate (Hz)", fontsize=13)
ax.set_title("Firing rate across trials", fontsize=15)
ax.tick_params(labelsize=12)
ax.legend(fontsize=11)
plt.tight_layout()
plt.show()
```

That line in the middle, `[firing_rate(c, duration_s) for c in spike_counts]`, is a list comprehension. It is a `for` loop squeezed into one line, and you will see it everywhere in Python code.

## Try it yourself

Change the bursting threshold in `classify()` from 20 to 8 and rerun the loop. Watch how many neurons change label. That is how sensitive threshold-based claims are, which is worth remembering when you read a paper.
