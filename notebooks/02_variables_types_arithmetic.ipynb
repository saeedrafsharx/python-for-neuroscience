## Why this matters

Everything else in this course is built on this one idea: a name that points to a value. A recording, a spike count, a subject ID, a threshold, all of them are just labelled values sitting in memory.

Ten minutes here saves you a lot of confusion later, so let's be precise about it.

## Variables

A variable is a name you attach to a value with `=`. Read it as "gets", not as "equals".

```python
firing_rate = 12.5          # spikes per second
neuron_id = "V1_042"        # which neuron this is
n_spikes = 250              # how many spikes we counted
is_responsive = True        # did it respond to the stimulus

print(firing_rate)
print(neuron_id)
```

## Types

Python cares what kind of thing a value is. There are four you will use constantly.

- `int`, a whole number. Spike counts, trial numbers, subject numbers.
- `float`, a decimal number. Firing rates, voltages, reaction times.
- `str`, text. Region names, condition labels, file paths.
- `bool`, either `True` or `False`. Did the neuron fire, did the subject respond correctly.

`type()` tells you which one you have.

```python
print(type(n_spikes))
print(type(firing_rate))
print(type(neuron_id))
print(type(is_responsive))
```

This matters in a very practical way. `"5"` and `5` look identical when printed and behave completely differently.

```python
print(5 + 5)        # arithmetic
print("5" + "5")    # text glued together
```

## Arithmetic, with a real calculation

Firing rate is spike count divided by recording duration. That is a one-liner, and it is a genuine piece of analysis.

```python
n_spikes = 250
duration_s = 20.0

rate = n_spikes / duration_s
print(rate)
```

The operators you need are `+`, `-`, `*`, `/`, and `**` for powers. Python also gives you `//` for integer division and `%` for the remainder, which is handy when you want to know things like "which trial block is trial 37 in".

```python
print(7 / 2)     # 3.5, ordinary division
print(7 // 2)    # 3, whole part only
print(7 % 2)     # 1, what is left over
print(2 ** 10)   # 1024
```

## Printing something readable

`print(rate)` gives you a bare number with no context. An f-string lets you put variables straight inside text, and it is what you will use for the rest of your life.

Put an `f` before the quote, and wrap any variable in curly braces.

```python
print(f"Neuron {neuron_id} fired {n_spikes} times in {duration_s} s")
print(f"That is a firing rate of {rate} Hz")
print(f"Rounded: {rate:.2f} Hz")     # .2f means two decimal places
```

## Putting it together

Four toy neurons, four firing rates, one figure. This is the first plot you make yourself.

```python
import matplotlib.pyplot as plt

# four neurons, spike counts from a 20 second recording
duration_s = 20.0
v1_spikes = 250
v2_spikes = 130
hippocampus_spikes = 44
motor_spikes = 310

names = ["V1", "V2", "Hippocampus", "Motor"]
rates = [
    v1_spikes / duration_s,
    v2_spikes / duration_s,
    hippocampus_spikes / duration_s,
    motor_spikes / duration_s,
]

for name, rate in zip(names, rates):
    print(f"{name:12s} {rate:5.1f} Hz")

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.bar(names, rates, color="#2b6cb0")
ax.set_ylabel("firing rate (Hz)", fontsize=13)
ax.set_title("Firing rate of four toy neurons", fontsize=15)
ax.tick_params(labelsize=12)
plt.tight_layout()
plt.show()
```
